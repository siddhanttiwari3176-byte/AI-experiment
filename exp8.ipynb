```python
import collections

def water_jug(capacity1, capacity2, target):
  visited = set()
  queue = collections.deque()
  queue.append((0, 0, []))

  while queue:
    jug1, jug2, path = queue.popleft()

    if (jug1, jug2) in visited:
      continue
    visited.add((jug1, jug2))

    current_path = path + [(jug1, jug2)]

    if jug1 == target or jug2 == target:
      return current_path

    next_states = [
        (capacity1, jug2),
        (jug1, capacity2),
        (0, jug2),
        (jug1, 0),
        (jug1 - min(jug1, capacity2 - jug2), jug2 + min(jug1, capacity2 - jug2)),
        (jug1 + min(jug2, capacity1 - jug1), jug2 - min(jug2, capacity1 - jug1))
    ]

    for state in next_states:
      if state not in visited:
          queue.append((state[0], state[1], current_path))

  return None



jug1_capacity = int(input("enter the capacity of jug1:"))
jug2_capacity = int(input("enter the capacity of jug2:"))
target_amount = int(input("enter the target:"))

solution = water_jug(jug1_capacity, jug2_capacity, target_amount)
if solution:
  print("solution path:")
  for state in solution:
    print(state)
else:
  print("no solution found")
```
